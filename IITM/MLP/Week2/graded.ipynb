Imputing age

```python
X_train[['Age']].isna().sum(), X_test[['Age']].isna().sum()
```

```python
age_imputer = SimpleImputer(strategy = 'mean')

age_imputer.fit(X_train[['Age']])

X_train[['Age']] = age_imputer.transform(X_train[['Age']])
X_test[['Age']] = age_imputer.transform(X_test[['Age']])
```

```python
X_train[['Age']].isna().sum(), X_test[['Age']].isna().sum()
```

```python
X_test['Age'].sum()
```

Imputing location

```python
X_train[['Location']].isna().sum(), X_test[['Location']].isna().sum()
```

```python
location_imputer = SimpleImputer(strategy = 'constant', fill_value = 'other')

location_imputer.fit(X_train[['Location']])

X_train[['Location']] = location_imputer.transform(X_train[['Location']])
X_test[['Location']] = location_imputer.transform(X_test[['Location']])
```

```python
X_train[['Location']].isna().sum(), X_test[['Location']].isna().sum()
```

Imputing Game Difficulty

```python
X_train[['GameDifficulty']].isna().sum(), X_test[['GameDifficulty']].isna().sum()
```

```python
difficulty_imputer = SimpleImputer(strategy = 'most_frequent')

difficulty_imputer.fit(X_train[['GameDifficulty']])

X_train[['GameDifficulty']] = difficulty_imputer.transform(X_train[['GameDifficulty']])
X_test[['GameDifficulty']] = difficulty_imputer.transform(X_test[['GameDifficulty']])
```

```python
X_train[['GameDifficulty']].isna().sum(), X_test[['GameDifficulty']].isna().sum()
```

Imputing Purchases

```python
X_train[['InGamePurchases']].isna().sum(), X_test[['InGamePurchases']].isna().sum()
```

```python
purchase_imputer = SimpleImputer(strategy = 'constant', fill_value = 0)

purchase_imputer.fit(X_train[['InGamePurchases']])

X_train[['InGamePurchases']] = purchase_imputer.transform(X_train[['InGamePurchases']])
X_test[['InGamePurchases']] = purchase_imputer.transform(X_test[['InGamePurchases']])
```

```python
X_train[['InGamePurchases']].isna().sum(), X_test[['InGamePurchases']].isna().sum()
```

```python
gender_ohe = OneHotEncoder(drop = 'first', sparse_output = False)

gender_ohe.fit(X_train[['Gender', 'Location', 'GameGenre']])

nominal_features_train = pd.DataFrame(gender_ohe.transform(X_train[['Gender', 'Location', 'GameGenre']]), columns=gender_ohe.get_feature_names_out())
nominal_features_test = pd.DataFrame(gender_ohe.transform(X_test[['Gender', 'Location', 'GameGenre']]), columns=gender_ohe.get_feature_names_out())
# nominal_features_train = pd.DataFrame(gender_ohe.transform(X_train[['Gender', 'Location', 'GameGenre']]).toarray(), columns=gender_ohe.get_feature_names_out())
# nominal_features_test = pd.DataFrame(gender_ohe.transform(X_test[['Gender', 'Location', 'GameGenre']]).toarray(), columns=gender_ohe.get_feature_names_out())

X_train.drop(columns = ['Gender', 'Location', 'GameGenre'], inplace = True)
X_test.drop(columns = ['Gender', 'Location', 'GameGenre'], inplace = True)
```
