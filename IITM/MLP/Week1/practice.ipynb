```python
def convert_to_seconds(row):
    '''
    This function will split the given time list into hours and minutes and return the time
    '''
    time_list = row['Duration'].split()
    total_seconds = 0
    
    # time is only in minutes or hours
    if len(time_list) == 1:
        if time_list[0][-1] == 'h':
            total_seconds += int(time_list[0][:-1])*60*60
        else: 
            total_seconds += int(time_list[0][:-1])*60
        return total_seconds
            
    else:    
        for ele in time_list:
            if ele[-1] == 'h':
                # an hour has 60 minutes, and 60 minutes have 60 seconds
                total_seconds += int(ele[:-1])*60*60
            else:
                # a minute has 60 seconds
                total_seconds += int(ele[:-1])*60
    return total_seconds
```

```python
# add an hour column
df["Duration in seconds"] = df.apply(convert_to_seconds, axis = 1)
```

```python
# Question 6

def dept_hour(time):
    x = time.split(":")
    hour = int(x[0])
    return hour

def arrival_hour(obj):
    x = obj.split(":")
    hour = int(x[0])
    return hour
```

```python
df["Dept_hour"] = df["Dep_Time"].apply(dept_hour)
df["Arrival_hour"] = df["Arrival_Time"].apply(arrival_hour)
```

```python
def cat_hour(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 20:
        return "Evening"
    elif 20 <= hour < 5:
        return "Night"
```

```python
df.sample()
```

```python
df['Cat_dept_hour'] = df["Dept_hour"].apply(cat_hour)
df['Cat_arrival_hour'] = df["Arrival_hour"].apply(cat_hour)
```

```python
# Question 7
def cat_weekday(day):
    if (day == "Saturday") or (day == "Sunday"):
        return 1
    else:
        return 0
```

```python
df["Cat_Week_Day"] = df["WeekDay"].apply(cat_weekday)
```
