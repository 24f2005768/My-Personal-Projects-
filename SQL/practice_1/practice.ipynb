```python
import sqlite3
import pandas as pd

import prettytable
# Set the default display format for prettytable to 'DEFAULT' (i.e., a simple table format)
prettytable.DEFAULT = 'DEFAULT'
```

```python
# Load data file
df1  = pd.read_csv("customers_medium.csv")
df2 = pd.read_csv("menu_items.csv")
df3 = pd.read_csv("order_items.csv")
df4 = pd.read_csv("orders_medium.csv")
df5 = pd.read_csv("restaurants.csv")
```

```python
# Connect to a sqlite db
conn = sqlite3.connect('practice.db')
cur = conn.cursor()
```

```python
%sql sqlite:///practice.db
```

```python
# Load data file to sqlite
# if exists: fail, replace, append
df1.to_sql("customers_medium", conn, if_exists = 'replace', index = False)
df2.to_sql("menu_items", conn, if_exists = 'replace', index = False)
df3.to_sql("order_items", conn, if_exists = 'replace', index = False)
df4.to_sql("orders_medium", conn, if_exists = 'replace', index = False)
df5.to_sql("restraunts", conn, if_exists = 'replace', index = False)
```

```python
# Get all tables of a database
%%sql
SELECT name 
FROM sqlite_master 
WHERE type='table';
```

```python
# close connection
conn.close()
```
