```python
import plotly.express as px
import plotly.graph_objects as go
```

## Fetching the Data

```python
df = px.data.gapminder().query("country == 'Canada'")
df.head()
```

### Simple plotly graph

```python
fig = px.line(df, x = 'year', 
              y = 'lifeExp', 
              title = "Life Expectancy in Canada",
              labels = {'lifeExp': 'Life Expectancy', 'year': 'Year'})

fig.show()
```

### Multiple lines Chart

```python
df_stocks = px.data.stocks()
df_stocks.head()
```

```python
px.line(df_stocks,
        x = 'date',
        y = ['GOOG', 'AAPL'],
        labels = {'date': 'Date', 'value': 'Price'},
        title = 'Google vs Apple')
```

### With GO

```python
fig = go.Figure()
fig.add_trace(go.Scatter(
    x = df_stocks.date,
    y = df_stocks.AAPL,
    mode = 'lines',
    name = 'Apple'
))
```
