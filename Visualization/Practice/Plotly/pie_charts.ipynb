```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

## Pie Chart

```python
data = {
    'Category': ['Product A', 'Product B', 'Product C', 'Product D'],
    'Value': [450, 300, 200, 50]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
fig = px.pie(
    df,
    values = 'Value',
    names = 'Category',
    title = 'Product Sales Distribution'
)

fig.show()
```

```python
df = px.data.gapminder().query("year == 2007").query("continent == 'Europe'")
```

```python
# group small countries together
df.loc[df['pop'] < 2.e6, 'country'] = 'Other Countries'
```

```python
fig = px.pie(
    df,
    values = 'pop',
    names = 'country',
    title = 'Population of European Continent (2007)'
)
fig
```

## With GO

```python
fig = go.Figure()
fig.add_trace(go.Pie(
    labels = ['Product A', 'Product B', 'Product C', 'Product D'],
    values = [450, 300, 200, 50],
    textinfo = 'label + percent',
    textposition = 'inside',
    )
)

fig.update_layout(
    title = 'Product Sales Distribution'
)
```

## Multiple Charts in one Figure

```python
fig = go.Figure()

# First pie chart
fig.add_trace(go.Pie(
    labels=['A', 'B', 'C'], 
    values=[40, 30, 30],
    name="Group 1",
    domain=dict(x=[0, 0.30])  # Left side
))

# Second pie chart
fig.add_trace(go.Pie(
    labels=['D', 'E', 'F'], 
    values=[25, 35, 40],
    name="Group 2",
    domain=dict(x=[0.70, 1]),  # Right side
    hole=0.4  # Donut style
))

fig.update_layout(
    title_text="Two Pie Charts Side by Side",
    showlegend=True
)
fig.show()
```

## Pull Slices

```python
# Method 1: Pull out specific slices
fig = go.Figure(data=[go.Pie(
    labels=['Product A', 'Product B', 'Product C', 'Product D'],
    values=[450, 300, 200, 50],
    pull=[0, 0.1, 0, 0],  # Pull out only Product B (index 1) by 0.1 radius
    textinfo='label+percent'
)])

fig.update_layout(title='Product B Emphasized')
fig.show()

# Method 2: Pull out multiple slices
fig2 = go.Figure(data=[go.Pie(
    labels=['A', 'B', 'C', 'D'],
    values=[40, 30, 20, 10],
    pull=[0, 0.1, 0.05, 0],  # Pull B and C by different amounts
    textinfo='label+percent'
)])
fig2.show()

# Method 3: Pull out one slice in Plotly Express
df = pd.DataFrame({'Category': ['A', 'B', 'C'], 'Value': [50, 30, 20]})
fig3 = px.pie(df, values='Value', names='Category', title='Emphasized Slice')
fig3.update_traces(pull=[0, 0.1, 0])  # Pull second slice
fig3.show()
```

## Multiple Charts and Subplots

```python
fig = go.Figure()

# Subplot 1 (top-left)
fig.add_trace(go.Pie(
    labels=['A', 'B', 'C'],
    values=[40, 35, 25],
    name="Dataset 1",
    domain=dict(x=[0, 0.45], y=[0.55, 1]),  # Top-left
    textinfo='label+percent'
))

# Subplot 2 (top-right)
fig.add_trace(go.Pie(
    labels=['D', 'E', 'F'],
    values=[30, 40, 30],
    name="Dataset 2",
    domain=dict(x=[0.55, 1], y=[0.55, 1]),  # Top-right
    textinfo='label+percent'
))

# Subplot 3 (bottom, centered)
fig.add_trace(go.Pie(
    labels=['G', 'H', 'I'],
    values=[50, 30, 20],
    name="Dataset 3",
    domain=dict(x=[0.25, 0.75], y=[0, 0.45]),  # Bottom center
    textinfo='label+percent'
))

fig.update_layout(
    title="Multiple Pie Charts",
    height=700,
    showlegend=False
)
fig.show()
```

## Hover

```python
df = pd.DataFrame({
    'Category': ['Product A', 'Product B', 'Product C'],
    'Sales': [50000, 75000, 35000],
    'Profit': [12000, 18000, 8000]
})

# Add custom hover data
fig = px.pie(df, values='Sales', names='Category', 
             title='Sales Distribution',
             hover_data={'Profit': ':$,.0f'})  # Format as currency

fig.update_traces(
    hovertemplate='<b>%{label}</b><br>Sales: $%{value:,.0f}<br>Profit: $%{customdata[0]:,.0f}<br>Percentage: %{percent}',
    textinfo='percent+label'
)
fig.show()
```

## Sort Values

```python
# Custom ordering with Graph Objects
fig = go.Figure(data=[go.Pie(
    labels=['Large', 'Medium', 'Small'],
    values=[50, 30, 20],
    sort=False  # Preserve your order
)])
fig.show()
```
