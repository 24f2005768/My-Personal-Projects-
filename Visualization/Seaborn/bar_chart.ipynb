```python
df = sns.load_dataset('titanic')
df.head()
```

```python
mart = pd.read_excel("supermarket_sales.xlsx")
mart.head()
```

```python
plt.figure(figsize = (15,5))
sns.barplot(x = 'Product line', y = 'Total', data = mart)
```

```python
# Add hue
plt.figure(figsize = (15,5))
sns.barplot(x = 'Product line', y = 'Total', data = mart, hue = "Gender")
```

```python
# Horizontal bar chart
plt.figure(figsize = (15,5))
sns.barplot(y = 'Product line', x = 'Total', data = mart)
```

```python
# plot in given order
plt.figure(figsize = (15,5))
sns.barplot(x='Product line',y='Total',hue='Gender',data=mart,
           hue_order=['Male','Female'],
           order = mart['Product line'].sort_values().unique()
)
```

```python
# add cap on error bar
plt.figure(figsize = (15,5))
sns.barplot(x='Product line',y='Total',hue='Gender',data=mart, capsize = 0.2)
```

```python
# Remove error bar
plt.figure(figsize = (15,5))
sns.barplot(x='Product line',y='Total',hue='Gender',data=mart, errorbar = None)
```

```python
# Change color 
plt.figure(figsize = (15,5))
sns.barplot(x='Product line',y='Total',hue='Gender',data=mart, errorbar = None, palette = "dark:pink")
```

```python
# change pallet
plt.figure(figsize = (15,5))
sns.barplot(x='Product line',y='Total',hue='Gender',data=mart, errorbar = None, palette = 'cividis')
```

```python
# Change saturation
plt.figure(figsize = (15,5))
sns.barplot(x='Product line',y='Total',hue='Gender',data=mart, errorbar = None, palette = 'cividis', saturation = 0.5)
```

```python
# default aggregation: default -> mean
plt.figure(figsize = (15,5))
sns.barplot(x='Product line',y='Total',hue='Gender',data=mart, errorbar = None, palette = 'cividis', estimator = sum)
```
