```python
import pandas as pd
import numpy as np

# ============================================
# 1. LOAD DATA
# ============================================
emissions_df = pd.read_csv('historical_emissions.csv')
country_code = pd.read_csv('country_code.csv', keep_default_na=False, na_values=[''])
country_df = pd.read_csv('world_country_and_usa_states_latitude_and_longitude_values.csv', 
                         keep_default_na=False, na_values=[''])

print("✅ Data loaded successfully")

# ============================================
# 2. CREATE SEPARATE MAPPING DICTIONARIES
# ============================================
# For coordinates dataset (country_df)
GEO_MAPPING = {
    'democratic republic of the congo': 'congo [drc]',
    'republic of congo': 'congo [republic]',
    'eswatini': 'swaziland',
    'macedonia': 'macedonia [fyrom]',
    'myanmar': 'myanmar [burma]',
}

# For country codes dataset (country_code)
CODE_MAPPING = {
    'brunei': 'brunei darussalam',
    'iran': 'iran, islamic republic of',
    'macedonia': 'macedonia, republic of',
    'micronesia': 'micronesia, federated states of',
    'north korea': 'korea (north)',
    'south korea': 'korea (south)',
    'republic of congo': 'congo (brazzaville)',
    'democratic republic of the congo': 'congo, (kinshasa)',
    'saint vincent and the grenadines': 'saint vincent and grenadines',
    'syria': 'syrian arab republic (syria)',
    'tanzania': 'tanzania, united republic of',
    'united states': 'united states of america',
    'venezuela': 'venezuela (bolivarian republic)',
    'vietnam': 'viet nam',
    'eswatini': 'swaziland',
    'russia': 'russian federation',
    'laos': 'lao pdr',
}

# ============================================
# 3. STANDARDIZE NAMES
# ============================================
# Create lowercase keys
emissions_df['_key'] = emissions_df['Country'].str.lower().str.strip()
country_df['_key'] = country_df['country'].str.lower().str.strip()
country_code['_key'] = country_code['Country_name'].str.lower().str.strip()

# Apply mappings
emissions_df['_geo_key'] = emissions_df['_key'].map(GEO_MAPPING).fillna(emissions_df['_key'])
emissions_df['_code_key'] = emissions_df['_key'].map(CODE_MAPPING).fillna(emissions_df['_key'])

# ============================================
# 4. MERGE DATA
# ============================================
# First merge: ISO codes
merged_df = emissions_df.merge(
    country_code[['_key', 'code_2digit', 'code_3digit']].rename(
        columns={'_key': '_lookup', 'code_2digit': 'iso2', 'code_3digit': 'iso3'}
    ),
    left_on='_code_key',
    right_on='_lookup',
    how='left'
).drop(columns=['_lookup'])

# Second merge: Coordinates
merged_df = merged_df.merge(
    country_df[['_key', 'latitude', 'longitude']].rename(columns={'_key': '_lookup'}),
    left_on='_geo_key',
    right_on='_lookup',
    how='left'
).drop(columns=['_lookup'])

# Clean up helper columns
merged_df = merged_df.drop(columns=['_key', '_geo_key', '_code_key'])

print("✅ Data merged successfully")

# ============================================
# 5. CHECK RESULTS FOR YOUR SPECIFIC COUNTRIES
# ============================================
# Check specific countries you mentioned
countries_to_check = [
    'World', 'United States', 'European Union (27)', 'Russia', 'Iran',
    'South Korea', 'Vietnam', 'Venezuela', 'Tanzania', 'Syria', 
    'Laos', 'North Korea', 'South Sudan', 'Brunei', 'Sao Tome and Principe',
    'Saint Vincent and the Grenadines'
]

print("\n📊 Checking specific countries:")
for country in countries_to_check:
    row = merged_df[merged_df['Country'] == country]
    if len(row) > 0:
        # FIX: Use pd.isna() instead of .isna() on individual values
        has_coords = not pd.isna(row.iloc[0]['latitude'])
        has_iso = not pd.isna(row.iloc[0]['iso2'])
        status = []
        if has_coords:
            status.append('📍')
        if has_iso:
            status.append('🏷️')
        status_str = ' '.join(status) if status else '❌'
        print(f"{status_str:6} {country}")
    else:
        print(f"❌ {country} (not found)")

# ============================================
# 6. SHOW COUNTRIES STILL MISSING COORDINATES
# ============================================
aggregates = {'world', 'european union (27)'}
missing = merged_df[
    merged_df['latitude'].isna() & 
    ~merged_df['Country'].str.lower().isin(aggregates)
]

print(f"\n📊 Missing coordinates summary:")
print(f"Total rows with missing coordinates: {merged_df['latitude'].isna().sum()}")
print(f"Countries missing coordinates (excluding aggregates): {missing['Country'].nunique()}")

if len(missing) > 0:
    print("\n⚠️ Countries still missing coordinates:")
    for country in missing['Country'].unique():
        print(f"  - {country}")

# ============================================
# 7. REORDER AND SAVE
# ============================================
year_cols = [col for col in merged_df.columns if col.isdigit()]
front_cols = ['Country', 'iso2', 'iso3', 'latitude', 'longitude']
merged_df = merged_df[front_cols + year_cols]

merged_df.to_csv('merged_country_co2_geo.csv', index=False)
print(f"\n✅ Saved to 'merged_country_co2_geo.csv'")

# Show first few rows
print("\n📋 Sample of merged data (first 5 rows):")
print(merged_df[['Country', 'iso2', 'iso3', 'latitude', 'longitude']].head())
```

```python
# ============================================
# ADD MANUAL COORDINATES FOR MISSING COUNTRIES
# ============================================
# Define manual coordinates for missing countries
manual_coords = {
    'South Sudan': (6.876991, 31.306978),
    'Sao Tome and Principe': (0.255436, 6.602781),
    'European Union (27)': (56.7000, 15.0000)
}

# Add coordinates to existing rows
for country, (lat, lon) in manual_coords.items():
    mask = merged_df['Country'] == country
    if mask.any():
        merged_df.loc[mask, 'latitude'] = lat
        merged_df.loc[mask, 'longitude'] = lon
        print(f"✅ Added coordinates for {country}: ({lat}, {lon})")
```
